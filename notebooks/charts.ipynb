```python
import pandas as pd
import matplotlib.pyplot as plt

df=pd.read_csv("../Data/employees_cleaned.csv")

dept_count = df.groupby('department')['id'].count()
avg_salary = df.groupby('department')['salary'].mean()
```

1. Bar Graph

```python
plt.figure(figsize=(8, 5))

plt.bar(dept_count.index, dept_count.values,
        color='skyblue', label='Employee Count')

plt.title('Number of Employees by Department')
plt.xlabel('Department')
plt.ylabel('Number of Employees')
plt.legend()

plt.savefig('bar_chart.png', dpi=300, bbox_inches='tight')
plt.show()
```

2. Histogram

```python
plt.figure(figsize=(8, 5))

plt.hist(df['salary'], bins=5, color='orange',
         edgecolor='black', label='Salary Distribution')

plt.title('Employee Salary Distribution')
plt.xlabel('Salary')
plt.ylabel('Frequency')
plt.legend()

plt.savefig('histogram.png', dpi=300, bbox_inches='tight')
plt.show()
```

3. Scatter Plot

```python
plt.figure(figsize=(8, 5))

plt.scatter(df['id'], df['salary'],
            color='green', label='Employees')

plt.title('Employee ID vs Salary')
plt.xlabel('Employee ID')
plt.ylabel('Salary')
plt.legend()

plt.savefig('scatter_plot.png', dpi=300, bbox_inches='tight')
plt.show()
```

4. Line Graph

```python
plt.figure(figsize=(8, 5))

plt.plot(avg_salary.index, avg_salary.values,
         marker='o', color='red', label='Average Salary')

plt.title('Average Salary by Department')
plt.xlabel('Department')
plt.ylabel('Average Salary')
plt.legend()

plt.savefig('line_chart.png', dpi=300, bbox_inches='tight')
plt.show()
```
